## 核心心智模型

### 1. 它是什么，解决什么问题

Dialect 是可组合 IR 词汇表；Type 描述 SSA 值的类别，Attribute 是编译期常量/配置，Interface 提供不依赖具体 op 名的能力契约。

### 2. 它如何工作

parser 根据 `dialect.op` 找定义；operation verifier 检查 operand/result/type/attribute 约束，interfaces 让 pass 对一组 op 统一工作。

### 3. 正确性条件与常见误区

不能把运行时 SSA 值塞进 attribute；未知 dialect 是否允许取决于工具注册与选项，验证通过也不代表语义已实现。

### 4. 性能与工程取舍

高层专用 type 保留语义、利于优化；过早降成 memref/LLVM pointer 会丢失 shape 与别名信息。
