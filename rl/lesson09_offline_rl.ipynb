## 核心心智模型

### 1. 它是什么，解决什么问题

离线 RL 只能使用固定数据；策略若选择数据外动作，Q 网络可能给出无依据的乐观外推。

### 2. 它如何工作

保守方法让数据分布外动作的价值更低，或把策略约束在行为策略附近；本质是为不确定性付费。

### 3. 正确性条件与常见误区

验证集 TD error 小不代表新策略安全，因为验证数据仍来自行为分布；必须分析 coverage。

### 4. 性能与工程取舍

约束太弱会利用 Q 误差，太强则退化为行为克隆，无法超过数据策略。
