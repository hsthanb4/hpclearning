## 核心心智模型

### 1. 它是什么，解决什么问题

DQN 用神经网络近似 Q，经验回放打破时序相关，目标网络减缓自举目标移动。

### 2. 它如何工作

采样转移后，以冻结目标网络计算 y=r+γmax_a Q_target(s',a)，在线网络回归该目标。

### 3. 正确性条件与常见误区

目标值必须 stop-gradient；终止时不能 bootstrap；回放分布与当前策略不同是 off-policy 前提而非 bug。

### 4. 性能与工程取舍

更慢的目标更新更稳但滞后；Double DQN 分离动作选择与动作评价来减轻 max 过估计。
